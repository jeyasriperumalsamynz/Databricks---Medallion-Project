```python
%sql
drop table if exists procat.proschema.silver_orders;

CREATE OR REPLACE TABLE procat.proschema.silver_orders
AS
SELECT
    CUSTOMER_ID                        AS customer_id,
    CUSTOMER_NAME                      AS customer_name,
    CUSTOMER_EMAIL                     AS customer_email,
    CITY                               AS city,
    REGION                             AS region,
    CAST(ORDER_DATE AS DATE)           AS order_date,
    CAST(UNIT_PRICE AS DECIMAL(12,2))  AS unit_price,
    CAST(QUANTITY AS INT)              AS quantity,
    CAST(UNIT_PRICE AS DECIMAL(12,2)) * CAST(QUANTITY AS INT) AS order_amount
FROM procat.proschema.bronze_orders;
select * from procat.proschema.silver_orders;
```
