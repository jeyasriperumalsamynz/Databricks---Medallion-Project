```python
%sql

drop table if exists procat.proschema.gold_orders;

CREATE TABLE if not exists procat.proschema.gold_orders (customer_id STRING, customer_name STRING, customer_email STRING, city STRING, region STRING, order_date DATE, unit_price DECIMAL(12,2), quantity INT, order_amount DECIMAL(14,2), order_year INT, order_month INT) PARTITIONED BY (order_year, order_month);


INSERT OVERWRITE TABLE procat.proschema.gold_orders PARTITION (order_year, order_month) SELECT customer_id, customer_name, customer_email, city, region, order_date, unit_price, quantity, order_amount, YEAR(order_date) AS order_year, MONTH(order_date) AS order_month FROM procat.proschema.silver_orders;


show partitions procat.proschema.gold_orders;
```
